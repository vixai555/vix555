```python
# 1. Load the file “zoo.data” and look at the info and the first five rows. 
# The first column denotes the animal name and the last one specifies a high-level class for the corresponding animal. 
# 2.  Find out the unique number of high-level cclasses 
# 3.  Use the 16-intermediate feature and perform an agglomerative clustering. [ Hint: Refer to the agglomerative clustering module in scikit learn and 
# set the number of clusters appropriately] 
# 4. Compute the mean squared error by comparing the actual class and predicted high-level class.
#4 The clustering does and cannot reproduce the original labels, only the original partitions. Hence #4 above is not possible.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn import preprocessing
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv("zoo.csv")
X = df.iloc[:,1:17]
y = df["class_type"]
print(len(set(y)))  # unique number of high-level classes 
```

```python
from sklearn.cluster import AgglomerativeClustering
agg = AgglomerativeClustering(n_clusters=5)
```

```python
clustered = agg.fit_predict(X)
```

```python
print(y, clustered)
```
