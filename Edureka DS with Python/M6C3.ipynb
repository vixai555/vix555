```python
# 3. Compute -- Explore types of relationships across the entire data set using pairplot. Based on this plot what looks to be the most correlated 
# feature with the Yearly Amount Spent? 

# Seaborn Pairplot is a Python library that allows you to plot pairwise relationships within a data set, 
# making it easier to visualize and understand large data sets. 

import seaborn as sns
import matplotlib. pyplot as plt
import pandas as pd

df = pd.read_csv("FyntraCustomerData.csv")
sns.pairplot(df)
plt.show()
```

```python
# 4. Compute – Create a linear model plot of the Length of Membership and Yearly Amount Spent. Does the data fit well in a linear plot? 
# 5. Compute – Train and Test the data and answer multiple questions -- What is the use of random_state=85? 
# 6. Compute – Predict the data and do a scatter plot. Check if actual and predicted data match. 
# 7. What is the value of Root Mean Squared Error?

cols_to_plot = df[['Yearly_Amount_Spent','Length_of_Membership']]
sns.pairplot(cols_to_plot, kind='reg', diag_kind='kde')
plt.show()
```

```python
# 5. Compute – Train and Test the data and answer multiple questions -- What is the use of random_state=85? 
# 6. Compute – Predict the data and do a scatter plot. Check if actual and predicted data match. 
# 7. What is the value of Root Mean Squared Error?

x = df.iloc[:,3:7]   # variables affecting the label - leaving out address and other strings because they may not affect the label
y = df["Yearly_Amount_Spent"] # label for prediction
print(x)
print(y)
```

```python
# 5. Compute – Train and Test the data and answer multiple questions -- What is the use of random_state=85? 
# 6. Compute – Predict the data and do a scatter plot. Check if actual and predicted data match. 
# 7. What is the value of Root Mean Squared Error?

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.33, random_state =85)
#fitting our model to train and test
lm = LinearRegression()
model = lm.fit(x_train,y_train)
```

```python
# 6. Compute – Predict the data and do a scatter plot. Check if actual and predicted data match. 
# 7. What is the value of Root Mean Squared Error?

pred_y = lm.predict(x_test)
plt.scatter(y_test,pred_y)
plt.xlabel('Y Test')
plt.ylabel('Predicted Y')
```

```python
# 7. What is the value of Root Mean Squared Error?

from sklearn.metrics import root_mean_squared_error
r = root_mean_squared_error(y_test,pred_y)
print ("Roor Mean Squared Error ", r)
```
