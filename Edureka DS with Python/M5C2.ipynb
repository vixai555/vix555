```python
# 3. How much money could have been saved in the Year 2014 by stopping OverTimePay? 
# 4. Which are the top 5 common jobs in the Year 2014 and how much do they cost SFO? 
# 5. Who was the top earning employee across all the years?

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("Salaries.csv")
hm = df[df['Year']==2014]
salary = hm[['TotalPay','JobTitle']]
sal = salary.groupby(by="JobTitle")
s1 = sal["JobTitle"].count()
#print(s1)

news = pd.DataFrame({'JobTitle':s1.index, 'Count':s1.values})
n = news.sort_values(by=['Count'], ascending=False)
print(n.head(5))
```

```python
# 3. How much money could have been saved in the Year 2014 by stopping OverTimePay? 

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("Salaries.csv")
hm = df[df['Year']==2014]
salary = df[['TotalPay','OvertimePay']]
tp = salary['TotalPay'].sum()
ot = salary['OvertimePay'].sum()
saved = tp - ot
print("Would have saved :" + str(saved))
```
