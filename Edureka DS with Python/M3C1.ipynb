```python
# Write a program that will find all such numbers which are divisible by 7 but are not a multiple of 5, between 2000 and 3200 (both included). 
# The numbers obtained should be printed in a comma-separated sequence on a single line. 

for x in range(2000, 3200):
    if (x%7 == 0):
        if (x%5 !=0):
            print(x, end = ", ")
```

```python
# Write a program that takes 2 digits, X, Y as input and generates a 2-dimensional array. 
# The element value in the i-th row and j-th column of the array should be i*j. Note: i=0,1.., X-1; j=0,1,¡-Y-1. 

import numpy as np


x = int(input("Value of X : "))
y = int(input("Value of Y : "))

x = int(x)
y = int(y)
arr = np.zeros((x, y), dtype=int)

for a in range(x):
    for b in range(y):
        arr[a,b] = a*b

print(arr)
```

```python
#  Write a program that accepts a comma-separated sequence of words as input and prints the words in a 
# comma-separated sequence after sorting them alphabetically.  

i = int(input("Enter the number of words you want to input :"))
print (i)
li = []
for x in range (i):
    li.append(input("Word :"))
li.sort()
for x in li:
    print (x, end = ", ")
```

```python
# Write a program that accepts a sequence of whitespace-separated words as input and prints 
# the words after removing all duplicate words and sorting them alphanumerically.  

x = input("Enter any string :")
y = x.split(" ")
y.sort()
z = set(y)
print(z)
```

```python
# Write a program that accepts a sequence of comma separated 4 digit binary numbers as its input and then check whether they are divisible by 5 or not. 
# The numbers that are divisible by 5 are to be printed in a comma-separated sequence. 

s = str(input("Enter the sequence of comma separated 4 digit binary numbers :"))
print (s)
x = s.split(",")
for z in x:
    z = int(z)
    if (z%5 == 0):
        print(z, end = ", ")
```
