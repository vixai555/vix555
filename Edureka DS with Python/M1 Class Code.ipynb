Python Operators

```python
#Airthmetic Operators: +,-,*,/,//,%(modulo),**
print(45+89)
print(45*30)
print(78/13)

print(10/3)  #output is float
print(10//3)  #Floor Division:returns int part of the result
print(10%3) #returns remainder division

print(10**3)  #x^y: Power operator,Exponentiation Operator
```

```python
#Assignment Operators: =,+=,-=
name='Amit'
#var1=var2 #right to left assignment

age=45
age+=10  #age=age+10
print(age)
age-=5  #age=age-5
print(age)
#++,-- are not allowed
```

```python
print(round(45.23698,2))
```

```python
#Conditional/Relational Operators: <,>,<=,>=,==,!=(not equal)
var1=63
var2=78
print(var1>var2)
print(var1<var2)

print(var1==var2) # Equality check
```

```python
#Logical Operators: and,or,not
age=36
print(age>=18 and age<=40 )
```

```python
age=45
print(age>=18 or age<=40 )
```
