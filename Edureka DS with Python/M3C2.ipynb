```python
# Domain – Banking Marketing 
# 1. Read file bank-data.csv 2. Build a set of unique jobs 3. Read the input from the command line –profession 
# 4. Check if the profession is on the list 5. Print whether the client is eligible 

import pandas as pd

df = pd.read_csv("bank-data.csv")
jobs = df['job']

prof = input("Enter your professional :")
for job in jobs:
    if (prof == job):
        print("Client Eligible")
        break;
```

```python
# Compute max and min age for loan eligibility based on data in CSV file 
# Store max and min age in the dictionary 

age = df['age']

max_age = max(age)
min_age = min(age)

print("Max Age : " + str(max_age))
print("Min Age : " + str(min_age))

age_dict = {
  "Max Age": max_age,
  "Min Age": min_age,
}

print(age_dict)
```
