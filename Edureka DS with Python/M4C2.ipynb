```python
#1. Read the three CSV files which contain the score of the same students in term1 of each Subject 

import pandas as pd
import re
import numpy as np

mf = pd.read_csv("MathScoreTerm1.csv")
df = pd.read_csv("DSScoreTerm1.csv")
pf = pd.read_csv("PhysicsScoreTerm1.csv")
```

```python
#2. Remove the name and ethnicity column (to ensure confidentiality)

m = mf[['Score','Age','Subject','Sex','ID']]
d = df[['Score','Age','Subject','Sex','ID']]
p = pf[['Score','Age','Subject','Sex','ID']]

print(p.head())
```

```python
#3 Fill missing score data with zero

m['Score'] = m['Score'].replace('',np.nan).fillna(0)
d['Score'] = d['Score'].replace('',np.nan).fillna(0)
p['Score'] = p['Score'].replace('',np.nan).fillna(0)
```

```python
# Change Sex(M/F) Column to 1/2 for further analysis

from sklearn.preprocessing import LabelEncoder
import pandas as pd

label_encoder = LabelEncoder()
cm = m['Sex']
encoded_data = label_encoder.fit_transform(cm)
m['Sex'] = encoded_data

cd = d['Sex']
encoded_data = label_encoder.fit_transform(cd)
d['Sex'] = encoded_data

cp = d['Sex']
encoded_data = label_encoder.fit_transform(cp)
d['Sex'] = encoded_data
```

```python
#4 4. Merge the three files
#6  Store the data in a new file – ScoreFinal.csv 
newdf = pd.read_csv("MathScoreTerm1.csv", usecols = ['ID','Age','Sex'])

math = m['Score']
phys = p['Score']
dstr = d['Score']

newdf['Maths'] = m['Score']
newdf['Physics'] = p['Score']
newdf['DS'] = d['Score']
print(newdf.head(5))

from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()
cm = newdf['Sex']
encoded_data = label_encoder.fit_transform(cm)
newdf['Sex'] = encoded_data

print(newdf.head(5))
newdf.to_csv("ScoreFinal.csv")
```
