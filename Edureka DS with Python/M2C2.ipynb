```python
#3 A website requires a user to input a username and password to register. Write a program to check the validity of the password given by the user. 
#Following are the criteria for checking password:   
#1. At least 1 letter between [a-z] 2. At least 1 number between [0-9]  1. At least 1 letter between [A-Z]  
#3. At least 1 character from [$#@]  4. Minimum length of transaction password: 6  5. Maximum length of transaction password: 12 

import re
def check_password_strength(s):
    # Check for at least one uppercase letter
    if not re.search(r"[A-Z]", s):
        return False
    # Check for at least one lowercase letter
    if not re.search(r"[a-z]", s):
        return False
    # Check for at least one digit
    if not re.search(r"[0-9]", s):
        return False
    # Check for at least one special character
    if not re.search(r"[!@#$%^&*()]", s):
        return False
    if (len(s)<6 or len(s)>12):
        return False
    return True

s = input("Password :")
x = check_password_strength(s)
if (x==True):
    print("Password Accepted")
else:
    print("Password Not Accepted")
```

```python
#5 Please write a program that accepts a string from the console and print the characters that have even indexes. 

s = input("Any text :")
x = len(s)

for a in range(x):
    if (a%2==0):
        print(s[a], end="")
```

```python
#6 Please write a program that accepts a string from the console and print it in reverse order. 

s = input("Any text :")
x = len(s)
y=x-1

for a in range(x):
    print(s[y], end="")
    y = y - 1
```

```python
#12 Write a program to compute 1/2+2/3+3/4+...+n/n+1 with a given n input by console (n>0). 


n = int(input("Any number :"))
y=0.0

for x in range(n+1):
    if (x!=0):
        y = y + (x/(x+1))
print(y)
```

```python
#11 With two given lists [1,3,6,78,35,55] and [12,24,35,24,88,120,155], write a program to make a list 
# whose elements are intersection of the above given lists. 

l1= [12,24,35,70,88,120,155]
l2= []
for i in range(len(l1)):
    if(l1[i]%5!=0 and l1[i]%7!=0):
        l2.append(l1[i])
print(l2)
```

```python
#10 By using list comprehension, please write a program to print the list after removing the 0th,4th, and 5th numbers in [12,24,35,70,88,120,155]. 

l1 = [12,24,35,70,88,120,155]
l2 = []
for i in range(len(l1)):
    if(i!=0 and i!=4 and i!=5):
        l2.append(l1[i])
print(l2)
```
