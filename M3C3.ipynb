```python
# 1. Read FairDealCustomerData.csv 2. Name field contains full name – use a regular expression to separate title, first name, last name 
# 3. Store the data in Customer Class 4. Create Custom Exception – CustomerNotAllowedException 
# 5. Pass a customer to function "createOrder" and throw CustomerNotAllowedException in case of blacklisted value is 1 

import pandas as pd
import re

df = pd.read_csv("FairDealCustomerData.csv")
names = df['name']
names_not_blacklisted = df.loc[df['blacklisted'] != 1, ['name', 'brand', 'blacklisted']]

def createOrder(n):
    df = pd.read_csv("FairDealCustomerData.csv")
    names_not_blacklisted = df.loc[df['blacklisted'] != 1, ['name']]
    print(names_not_blacklisted)
    for name in names_not_blacklisted:
        if (n == name):
            return True
        else:
            return False

y = []
for name in names:
    name = str(name)
    x = re.split(" ", name)
    y.append(x)

print("Separated Names :")
for x in range(10):
    print(y[x])

name =  input("Customer Name :")
y = createOrder(name)
if y == True:
    print(name + " Not blacklisted")
else:
    print(name + " blacklisted")
```
