```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import xgboost as xgb
from sklearn.preprocessing import LabelEncoder as le
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
from sklearn.metrics import f1_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import average_precision_score
from sklearn.metrics import recall_score

import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('cust_predict_bill_pay.csv', delimiter=',', header=0, encoding='utf-8')
```

```python
y =  df["Last_Bill_Paid_on_Time"]
```

```python
x = df[['Age','Salary_monthly','Gender','Average_bill_value','Phone_calls_Last_month','Average_duration_phone_calls']]
```

```python
def cm(df,col):
    lc = le()
    c = df[col]
    encoded_data = lc.fit_transform(c)
    df[col] = encoded_data

for d in x:
    d1_data = x[d]
    t = type((d1_data[0]))
    if t is str:
        cm(x,d)
#print(x)
lc = le()
y = lc.fit_transform(y)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 42)

# Initialize XGBClassifier
xgb_clf = xgb.XGBClassifier(use_label_encoder=False, eval_metric='logloss')
# Fit the model
xgb_clf.fit(x_train, y_train)
y_pred = xgb_clf.predict(x_test)
acc = accuracy_score(y_test, y_pred)
print("Accuracy_XGB :")
print(f"{acc:.2%}")
cmt = confusion_matrix(y_test, y_pred)
tn, fp, fn, tp = cmt.ravel()
print("True Positives :", tp)
print("False Positives :", fp)
print("True Negatives :", tn)
print("False Negatives :", fn)
print("Precision :") # P = TP / (TP + FP)
ps = average_precision_score(y_test, y_pred)
print(f"{ps:.2%}")
print("Recall :") # R = TP / (TP + FN)
rs = recall_score(y_test, y_pred)
print(f"{rs:.2%}")
print("F1 Score :") # F1 = Harmonic Mean of P & R
f1 = f1_score(y_test, y_pred)
print(f"{f1:.2%}")
print("Confusion Matrix :")
print(cmt)
```
