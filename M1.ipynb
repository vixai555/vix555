```python
#1 Write a program that will find factors of the given number and find whether the factor is even or odd. 

def print_factors(x):
   print("The factors of",x,"are:")
   for i in range(1, x + 1):
       if x % i == 0:
           print(i, end = " ")
   print ("\n")
   for i in range(1, x + 1):
       if (x % i == 0):
        print (i)
        if (i%2)==1:
            print (" Odd")
        else:
            print(" Even")

num = int(input("Enter any number :"))
print_factors(num)
```

```python
#3 Write a program, which will find all the numbers between 1000 and 3000 (both included) such that each digit of a number is an even number. 
# The numbers obtained should be printed in a comma-separated sequence on a single line. 

for n in range (1000, 3000, 1):
    n0 = n// 10**0 % 10
    n1 = n// 10**1 % 10
    n2 = n// 10**2 % 10
    n3 = n// 10**3 % 10
    if (n0%2==0 and n1%2==0 and n2%2==0 and n3%2==0):
        print (n3, ",", n2, ",", n1,",", n0)
```

```python
#4 Write a program that accepts a sentence and calculates the number of letters and digits. 

l = 0
d = 0
s = input("Any sentence with words & numbers :")
# Iterate through each character in the string
for char in s:
  # Check if the character is a letter
    if char.isalpha():  
        l = l + 1
   # Check if the character is a digit
    else: 
        d = d + 1
print(f"Letters: {l}")
print(f"Digits: {d}")
```

```python
#5  Design a code that will find whether the given number is a Palindrome number or not. 

s = input("Any potential palidrome string/number :")
rev = ""  # to store the reversed string
for char in s:
    rev = char + rev
if s == rev:
    print("Yes")
else:
    print("No")
```
