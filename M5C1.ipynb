```python
# 2. Find the genre in which there has been the greatest number of movie releases 

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

h = pd.read_csv("HollywoodMovies.csv")
hm = h[['Genre','Movie']]
n = hm.groupby(["Genre"])['Movie'].count()
x = n.index.values
y = n

plt.bar(x, y, color = "green")
plt.figure(figsize=(20,6))
plt.show()
```

```python
#5 has already been completed as part of earlier case study

#6 Perform Operations on Files  6.1: From the raw data below create a Pandas Series 'Amit', 'Bob', 'Kate', 'A', 'b', np.nan, 'Car', 'dog', 'cat' 
# a) Print all elements in lower case b) Print all the elements in upper case  c) Print the length of all the elements 

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = ('Amit', 'Bob', 'Kate', 'A', 'b', np.nan, 'Car', 'dog', 'cat')
s1 = pd.Series(data)

try:
    for s in s1:
        print(s.upper())
        print(s.lower())
        print(len(s))
except:
    print("An exception occurred")
```

```python
#6 6.3: -  Create a series from the raw data below 'India_is_big', 'Population_is_huge', np.nan, 'Has_diverse_culture' 
# a) split the individual strings wherever ‘_’ comes and create a list out of it. b) Access the individual elements of a list 
#c) Expand the elements so that all individual elements get split by ‘_’ and instead of list returns individual elements 

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = ('India_is_big', 'Population_is_huge', 'Has_diverse_culture' )
s1 = pd.Series(data)

l = []
for s in s1:
    index = s.find("_")
    if index != -1:
        a = s.split("_")
        l.append(a)
print(l)
```

```python
# 6.8: Create pandas series and print true if the value is containing ‘A’ '1', '2', '1a', '2b', 'America', 'VietnAm','vietnam', '2003c' 

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = ('1', '2', '1a', '2b', 'America', 'VietnAm','vietnam', '2003c' )
s1 = pd.Series(data)

for s in s1:
    index = s.find("A")
    if index != -1:
        print(s)
```
